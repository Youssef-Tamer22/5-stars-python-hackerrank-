## 13. Lists (Easy)

```python
if __name__ == '__main__':
    arr=[]   
    N = int(input())
    for _ in range(N):
        cmd = input().lower().split()
        if cmd[0]=='insert': 
            arr.insert(int(cmd[1]),int(cmd[2]))
        elif cmd[0]=='remove':
            arr.remove(int(cmd[1]))
        elif cmd[0]=='append':
            arr.append(int(cmd[1]))
        elif cmd[0]=='sort':
            arr.sort()
        elif cmd[0]=='pop':
            arr.pop()
        elif cmd[0]=='reverse':
            arr.reverse()
        elif cmd[0]=='print':
            print(arr)  
```

## 15. Nested Lists (Easy)

```python
if __name__ == '__main__':
    students =[]
    for _ in range(int(input())):
        name = input()
        score = float(input())
        students.append([name,score])
    grades = sorted(list(set(student[1] for student in students)))
    second_lowest = grades[1]
    second_lowest_student = sorted(student[0] for student in students if student[1] == second_lowest)
    for name in second_lowest_student:
        print(name)
```

## 16. Finding the percentage (Easy)

```python
if __name__ == '__main__':
    n = int(input())
    student_marks = {}
    for _ in range(n):
        name, *line = input().split()
        scores = list(map(float, line))
        student_marks[name] = scores
    query_name = input()

average=sum(student_marks[query_name])/3
print(f'{average:.2f}')
```

## 17. Find a string (Easy)

```python
def count_substring(string, sub_string):
    counter=0
    for i in range(0,len(string)):
        if string[i : i + len(sub_string)] == sub_string:
            counter += 1
    return counter

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()
    count = count_substring(string, sub_string)
    print(count)
```

## 20. The Minion Game (Medium)

```python
def minion_game(string):
    kevin=0
    stuart=0
    vowel={'A','E','I','O','U'}
    for i in range(len(string)):
        if string[i] in vowel:
            kevin+=len(string)-i
        else:
            stuart+=len(string)-i
    if kevin>stuart:
        print(f"Kevin {kevin}")
    elif stuart>kevin:
        print(f'Stuart {stuart}')
    else :
        print('Draw')

if __name__ == '__main__':
    s = input()
    minion_game(s)
```

## 23. Merge the Tools! (Medium)

```python
def merge_the_tools(string, k):
    count=0
    s=""
    for ch in string:
        if count==k:
            print(s)
            s=""
            count=0
        if ch not in s:
            s+=ch
        count+=1
    print(s)

if __name__ == '__main__':
    string, k = input(), int(input())
    merge_the_tools(string, k)
```
